# Setup

Everyone run all cells, then switch kernel to `Python (dropout-prediction)`.

```python
ENV_NAME = "dropout-prediction"
PYTHON_VERSION = "3.14"
```

```python
import subprocess

def run(*args):
    r = subprocess.run(list(args), capture_output=True, text=True)
    print(r.stdout[-1500:])
    if r.returncode != 0:
        print(r.stderr[-1500:])
        raise RuntimeError(" ".join(args))
    return r

envs = subprocess.run(["conda", "env", "list"], capture_output=True, text=True).stdout
if ENV_NAME not in envs:
    run("conda", "create", "-y", "-n", ENV_NAME, f"python={PYTHON_VERSION}")

run("conda", "run", "-n", ENV_NAME, "pip", "install", "-r", "requirements.txt", "ipykernel")
run("conda", "run", "-n", ENV_NAME, "python", "-m", "ipykernel", "install",
    "--user", "--name", ENV_NAME, "--display-name", f"Python ({ENV_NAME})")

print("Done. Switch kernel to:", f"Python ({ENV_NAME})")
```
